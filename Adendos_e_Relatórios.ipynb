```python
import pandas as pd
import sqlite3
import re
```

```python
db_path = '/content/drive/MyDrive/PRIVATE_DATA/Relatorio_de_Alta_Completos.db'
conn = sqlite3.connect(db_path)
```

```python
# Substitua com o nome real da sua tabela
query = 'SELECT * FROM relatório_de_alta'

# Execute a consulta e obtenha os resultados como um DataFrame
df = pd.read_sql(query, conn)


def processar_texto(texto):
    palavras = texto.split()
    if palavras and palavras[0] == 'ETIQUETA':
        try:
            indice_rn = palavras.index('RN')
            return ' '.join(palavras[indice_rn:])
        except ValueError:
            # 'RN' não encontrado, manter o texto original
            pass
    return texto

# Aplicar a função à coluna 'Texto Completo'
df['Texto Completo'] = df['Texto Completo'].apply(processar_texto)

# Exibindo o DataFrame após o processamento
display(df)
```

```python
df['Texto sem Espaço'] = df['Texto Completo'].apply(lambda x: x.replace(" ", ""))
df
```

```python
def remover_entre_etiqueta_e_nomedaM(texto):
    # Defina o padrão da expressão regular
    padrao = r'ETIQUETA.*?NomedaM'

    # Use a função sub para substituir o padrão por uma string vazia
    texto = re.sub(padrao, '', texto)

    return texto



################################################################
def extrair_data_nascimento(texto):
    # Defina o padrão da expressão regular
    padrao =  r'DatadeNascimento(.{11})'

    # Use a função findall para encontrar todas as correspondências no texto
    correspondencias = re.findall(padrao, texto, re.IGNORECASE)

    # Verifique se houve correspondências
    if correspondencias:
        # Retorna a primeira correspondência encontrada (os 10 caracteres)
        return correspondencias[0]

    # Retorna None se não houver correspondências
    return None

#####################################################################################################
####SEXO
def extrair_sexo(texto):
    # Defina o padrão da expressão regular
    padrao = r'Sexo:\s*(\w{3})'

    # Use a função findall para encontrar todas as correspondências no texto
    correspondencias = re.findall(padrao, texto, re.IGNORECASE)

    # Verifique se houve correspondências
    if correspondencias:
        # Retorna a primeira correspondência encontrada (os 3 caracteres após Sexo:)
        return correspondencias[0]

    # Retorna None se não houver correspondências
    return None

##############################################################################################

def extrair_nome_da_mae(texto):
    padrao = re.compile(r'RN(.*?)Data', re.DOTALL)
    resultado = padrao.search(texto)

    if resultado:
        informacoes = resultado.group(1).strip()
        return informacoes
    else:
        return None

##############################################################################################

def extrair_GBS(texto):
    # Defina o padrão da expressão regular
    padrao = r'GBS(.*?)Tipagem'

    # Use a função findall para encontrar todas as correspondências no texto
    correspondencias = re.findall(padrao, texto, re.DOTALL)

    # Verifique se houve correspondências
    if correspondencias:
        # Retorna a primeira correspondência encontrada (o conteúdo entre GBS e Tipagem)
        return correspondencias[0].strip()

    # Retorna None se não houver correspondências
    return None

###################################################################
def extrair_tipo_sangue(texto):
    padrao = r"(?:Sanguínea:)(.*?)(?:Intercorrências)"
    resultado = re.search(padrao, texto)

    if resultado:
        conteudo_entre = resultado.group(1).strip()
        return conteudo_entre
    else:
        return None

#####################################################################
def extrair_comprimento(texto):
    # Defina o padrão da expressão regular
    padrao = r'Comprimento:\s*([\d,.]+)'

    # Use a função findall para encontrar todas as correspondências no texto
    correspondencias = re.findall(padrao, texto)

    # Verifique se houve correspondências
    if correspondencias:
        # Retorna a primeira correspondência encontrada (o número após Comprimento)
        return correspondencias[0]

    # Retorna None se não houver correspondências
    return None

#########################################################################
def extrair_PC(texto):
    # Defina o padrão da expressão regular
    padrao = r'PC:\s*([\d,.]+)'

    # Use a função findall para encontrar todas as correspondências no texto
    correspondencias = re.findall(padrao, texto)

    # Verifique se houve correspondências
    if correspondencias:
        # Retorna a primeira correspondência encontrada (o número após Comprimento)
        return correspondencias[0]

    # Retorna None se não houver correspondências
    return None
###########################################################################
def extrair_PT(texto):
    # Defina o padrão da expressão regular
    padrao = r'PT:\s*([\d,.]+)'

    # Use a função findall para encontrar todas as correspondências no texto
    correspondencias = re.findall(padrao, texto)

    # Verifique se houve correspondências
    if correspondencias:
        # Retorna a primeira correspondência encontrada (o número após Comprimento)
        return correspondencias[0]

    # Retorna None se não houver correspondências
    return None

############################################################################
def extrair_tipo_de_parto(texto):
    padrao = r"(?:PARTO:)(.*?)(?:Anestesia)"
    resultado = re.search(padrao, texto, re.DOTALL | re.IGNORECASE)

    if resultado:
        conteudo_entre = resultado.group(1).strip()
        return conteudo_entre
    else:
        return None

##############################################################################

def extrair_quantidade_pre_natal(texto):
    padrao = padrao = re.compile(r'Pré-Natal\D*(.*?)co', re.IGNORECASE | re.UNICODE)

    resultado = padrao.search(texto)

    if resultado:
        quantidade_pre_natal = resultado.group(1)[:5]  # Limita aos primeiros 5 caracteres
        return quantidade_pre_natal
    else:
        return None

#############################################################################


def extrair_tipo_de_parto(texto):
    padrao = r"PARTO:(.*?)(?:Anestesia)"
    resultado = re.search(padrao, texto, re.DOTALL)

    if resultado:
        conteudo_entre = resultado.group(1).strip()
        return conteudo_entre
    else:
        return None

#############################################################################################

def extrair_peso(texto):
    padrao = r"Peso:(.*?)(?:g)"
    resultado = re.search(padrao, texto, re.DOTALL | re.IGNORECASE)

    if resultado and resultado.group(1) is not None:
        peso_formatado = re.sub(r'[^\d.]', '', resultado.group(1))
        return peso_formatado
    else:
        return None
###################################################################

def extrair_apgar(texto):
    padrao = re.compile(r'\bApgar\b', re.IGNORECASE)
    resultado = padrao.search(texto)

    if resultado:
        return 1
    else:
        return 0

##############################################################################
def extrair_idade_gestacional(texto):
    padrao = re.compile(r'\bCapurro\b', re.IGNORECASE)
    resultado = padrao.search(texto)

    if resultado:
        return 1
    else:
        return 0

#############################################################################
def extrair_indicacao_cesaria(texto):
    palavras_chave = [
        "cardiotocografiacategoria3",
        "cardiotocografianãotranquilizadora",
        "cardiotocografiaanormal",
        "braquicardiafetal",
        "sofrimentofetal",
        "SF",
        "SFA",
        "SFC",
        "restriçãodecrescimento",
        "RCF",
        "RCIU",
        "PIG",
        "pequenoparaidadegestacional",
        "insuficiênciaplacentária"
    ]

    palavras_encontradas = []

    for palavra in palavras_chave:
        padrao = re.compile(rf'\b{palavra}\b', re.IGNORECASE | re.UNICODE)
        resultado = padrao.search(texto)

        if resultado:
            palavras_encontradas.append(resultado.group())

    return palavras_encontradas

##############################################################################
def extrair_diagnostico_internacao(texto):
    palavras_chave = [
        "PIG",
        "pequenoparaidadegestacional",
        "P05",
        "baixopeso",
        "muitobaixopeso",
        "extremobaixopeso",
        "BP",
        "BPN",
        "MBP",
        "P07",
        "infecçãointra-amniótica",
        "P39.2",
        "P39.9",
        "asfixiaperinatal",
        "asfixiaaonascimento",
        "P21",
        "acidosemetabólica",
        "encefalopatia hipóxica-isquêmica",
        "EHI",
        "P91",
        "Enterocolitenecrosante",
        "EN",
        "P77",
        "LeucomaláciaPeriventricular",
        "LPV",
        "P91",
        "criseconvulsiva",
        "convulsão",
        "P90",
        "hemorragiaintraventricular",
        "HIV",
        "HPIV",
        "hemorragiaperiintraventricular",
        "hemorragiaperintraventricular",
        "P52",
        "Displasica Broncopulmonar",
        "DBP",
        "P27",
        "Desconfortorespiratório",
        "SDR",
        "DRP",
        "desconfortorespiratórioproce",
        "P22",
        "P228",
        "P22.8",
        "NecessidadedeventilaçãoMecânica",
        "intubação",
        "intubacao",
        "intubaçao",
        "apneiadaprematuridade",
        "P283",
        "P28.3",
        "TTRN",
        "Taquipnéiatransitoria",
        "Taquipnéiatransitória",
        "Taquipneiatransitória",
        "Sepse",
        "P39.8",
        "P398",
        "P-398",
        "P-39.8",
        "P36",
        "P-36",
        "Hipoglicemianeonatal"
        "P70.4",
        "P704",
        "P-70.4",
        "P-704",
        "icterícianeonatal",
        "ictericianeonatal",
        "INN",
        "P59",
        "P59.9",
        "P-59.9",
        "recémnascidoprétermo",
        "recemnascidopretermo",
        "recémnascidopré-termo",
        "recemnascidopre-termo",
        "RNPT",
        "P07",
        "P-07",
    ]

    palavras_encontradas = []

    for palavra in palavras_chave:
        padrao = re.compile(rf'\b{palavra}\b', re.IGNORECASE | re.UNICODE)
        resultado = padrao.search(texto)

        if resultado:
            palavras_encontradas.append(resultado.group())

    return palavras_encontradas

#############################################################################
def extrair_conduta_inicial(texto):
    palavras_chave = [
        "suporteventilatorio",
        "suporteventilatório",
        "UTI",
        "UTIN",
        "UCI",
        "UCIN",
    ]

    palavras_encontradas = []

    for palavra in palavras_chave:
        padrao = re.compile(rf'\b{palavra}\b', re.IGNORECASE | re.UNICODE)
        resultado = padrao.findall(texto)

        if resultado:
            palavras_encontradas.extend(resultado)

    return palavras_encontradas



```

```python
import os
dados_adendo = []


for index, row in df.iterrows():
    texto = row['Texto sem Espaço']
    texto2 = row['Texto Completo']
    texto = remover_entre_etiqueta_e_nomedaM(texto)
    arquivo = row['Nome do Arquivo']
    data_nascimento = extrair_data_nascimento(texto)
    sexo = extrair_sexo(texto)
    mae = extrair_nome_da_mae(texto2)
    GBS = extrair_GBS(texto)
    tipo_sangue = extrair_tipo_sangue(texto)
    comprimento = extrair_comprimento(texto)
    PC = extrair_PC(texto)
    PT = extrair_PT(texto)
    parto = extrair_tipo_de_parto(texto)
    pre_natal = extrair_quantidade_pre_natal(texto)
    tipo_parto = extrair_tipo_de_parto(texto)
    peso = extrair_peso(texto)
    apgar = extrair_apgar(texto)
    idade_gestacional = extrair_idade_gestacional(texto)
    indicacao_cesaria = extrair_indicacao_cesaria(texto)
    diagnostico_de_internacao = extrair_diagnostico_internacao(texto)
    conduta_inicial = extrair_conduta_inicial(texto)
    print(conduta_inicial)
    # Adicionando os dados ao DataFrame
    dados_adendo.append({
        "Nome do Arquivo": arquivo,
        "Texto": texto,
        "Peso":peso,
        "Quantidade Consultas Pré-Natal" : pre_natal,
        "Data de Nascimento": data_nascimento,    #OK
        "Sexo": sexo, #OK
        "Nome da Mãe": mae, #OK
      #  "GBS": GBS, #ok
        "Tipo de Sangue": tipo_sangue, #OK
        "Comprimento": comprimento, #OK
        "PC": PC, #OK
        "PT": PT, #OK
        "Tipo de Parto": tipo_parto,
        "Indicação de Cesária": indicacao_cesaria, #OK
        "Diagnostico de Internação":diagnostico_de_internacao, #OK
        "Conduta Inicial": conduta_inicial


    })

df_notas_clinicas = pd.DataFrame(dados_adendo)

pd.set_option('display.max_rows', 1000)  # Configura para exibir até 1000 linhas

# Exibindo as mil primeiras linhas do DataFrame
df_notas_clinicas.head(1000)
```

```python
#Formatando algumas coisas

df_notas_clinicas['Nome da Mãe'] = df_notas_clinicas['Nome da Mãe'].str.replace(r'^\s*[dD][eE]:?\s*', '', regex=True)

df_notas_clinicas['Peso'] = df_notas_clinicas['Peso'].str.replace(r'[.,]', '', regex=True)
df_notas_clinicas['Peso'] = df_notas_clinicas['Peso'].where(df_notas_clinicas['Peso'].str.len().between(3, 4))
df_notas_clinicas['Data de Nascimento'] = df_notas_clinicas['Data de Nascimento'].str.replace(r'[^0-9/]', '', regex=True)

def mapear_sexo(valor):
    if valor is not None:
        valor = valor.lower()  # Converte para minúsculas para lidar com variações de caixa
        if valor.startswith('m'):
            return 'M'
        elif valor.startswith('f'):
            return 'F'
    return valor

# Aplica a função à coluna 'Sexo'
df_notas_clinicas['Sexo'] = df_notas_clinicas['Sexo'].apply(mapear_sexo)

mascara = df_notas_clinicas['Quantidade Consultas Pré-Natal'].str.match(r'^\d{1,2}$', na=False)

# Substitui os valores que não atendem à condição por None (ou outro valor desejado)
df_notas_clinicas.loc[~mascara, 'Quantidade Consultas Pré-Natal'] = None


# Exemplo de uso
print(df_notas_clinicas['Quantidade Consultas Pré-Natal'])
```
